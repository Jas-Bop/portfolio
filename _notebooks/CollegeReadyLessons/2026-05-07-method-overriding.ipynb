<summary>Method Overriding</summary>

Method overriding happens when a child class provides its own version of a method inherited from a parent class. The method name stays the same, but the child behavior becomes more specific.

<pre><code class="language-javascript">class Alert {
  send() {
    console.log("Sending alert");
  }
}

class EmailAlert extends Alert {
  send() {
    console.log("Sending email alert");
  }
}</code></pre>
