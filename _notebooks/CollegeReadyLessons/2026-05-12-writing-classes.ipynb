<summary>Writing Classes</summary>

A class is a reusable blueprint that bundles related data and behavior together. The constructor sets up each object, while methods describe what the object can do. Classes are useful when a project needs several objects with the same structure but different values.

<pre><code class="language-javascript">class Student {
  constructor(name, grade) {
    this.name = name;
    this.grade = grade;
  }

  introduce() {
    console.log(this.name + " is in grade " + this.grade);
  }
}

const student = new Student("Ava", 9);
student.introduce();</code></pre>
