<summary>Inheritance</summary>

Inheritance lets a child class reuse behavior from a parent class. Shared methods stay in the parent, while child classes add specialized behavior. This keeps related classes organized without copying the same code repeatedly.

<pre><code class="language-javascript">class User {
  login() {
    console.log("User logged in");
  }
}

class Admin extends User {
  deletePost() {
    console.log("Post deleted");
  }
}</code></pre>
