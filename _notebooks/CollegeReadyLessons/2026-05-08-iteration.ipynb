<summary>Loops</summary>

Loops repeat code. They are useful when the same action needs to happen for every item in a list or for a known number of repetitions.

<pre><code class="language-javascript">const tasks = ["read", "practice", "submit"];

for (let i = 0; i &lt; tasks.length; i++) {
  console.log(tasks[i]);
