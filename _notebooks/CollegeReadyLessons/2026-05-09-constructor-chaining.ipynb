<summary>Constructors and super()</summary>

Constructor chaining sets up parent and child class data in the correct order. A child class uses super() to run the parent constructor before assigning its own properties.

<pre><code class="language-javascript">class Account {
  constructor(username) {
    this.username = username;
  }
}

class TeacherAccount extends Account {
  constructor(username, subject) {
    super(username);
    this.subject = subject;
  }
}</code></pre>
