<summary>Instantiation and Objects</summary>

Instantiation creates a real object from a class blueprint. Each instance follows the same class structure but stores its own data. This lets one class create many independent objects.

<pre><code class="language-javascript">class Course {
  constructor(name, period) {
    this.name = name;
    this.period = period;
  }
}

const course = new Course("JavaScript", 2);</code></pre>
