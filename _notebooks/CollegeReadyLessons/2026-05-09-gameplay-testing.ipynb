## Gameplay Testing Lesson

<p style="font-style:italic;">The lesson starts from a normal programming situation and keeps the example focused. Every line in the sample is there to show the concept doing work.</p>

Gameplay testing checks whether a game behaves correctly while someone actually plays it. It includes controls, movement, scoring, collisions, difficulty, restarts, win states, and failure states.

This kind of testing is different from only reading the code. A feature can look correct in the file but feel wrong during play. Movement might be too fast, collisions might be too large, or a restart might leave old state behind.



```python
%%js

// CODE_RUNNER: Gameplay Testing Lesson

function checkLevelComplete(player, level) {
  if (player.score >= level.targetScore) {
    level.complete = true;
    console.log("Level complete");
  }

  return level.complete;
}
```



The function checks a playable goal: whether the score is high enough to finish the level. Gameplay testing would confirm that this happens at the right time during play.
