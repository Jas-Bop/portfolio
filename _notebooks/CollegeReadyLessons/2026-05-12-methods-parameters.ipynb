<div style="border-top: 5px solid #2563eb; background:#eff6ff; padding:1rem;">
<strong>Method</strong>
<p>A method is an action stored inside an object or class. It is written like a function, but it belongs to the object that uses it.</p>
</div>

<div style="border-top: 5px solid #16a34a; background:#f0fdf4; padding:1rem;">
<strong>Parameter</strong>
<p>A parameter is the placeholder name inside the method definition. It tells the method what kind of input it expects to receive.</p>
</div>

<div style="border-top: 5px solid #dc2626; background:#fef2f2; padding:1rem;">
<strong>Argument</strong>
<p>An argument is the real value sent into the method when the method is called. The argument fills the parameter for that run.</p>
</div>

</div>

A method becomes useful when it can do the same kind of job with different values. In the scoreboard example, the action is adding points. The method does not need to know ahead of time whether the player earned 10 points, 25 points, or some other amount. The parameter gives the method a flexible opening for that value.



```python
%%js

// CODE_RUNNER: Methods and Parameters Lesson

class Scoreboard {
  constructor() {
    this.score = 0;
  }

  addPoints(points) {
    this.score = this.score + points;
    console.log("Score: " + this.score);
  }
}

const board = new Scoreboard();
board.addPoints(10);
board.addPoints(25);
```


<div style="border-left: 6px solid #9333ea; padding-left: 1rem; margin-top: 1rem;">

The method is <code>addPoints</code>. The parameter is <code>points</code>. The arguments are <code>10</code> and <code>25</code>, because those are the real values passed into the method calls. This lets one method update the score in multiple situations without copying the same code.

</div>
