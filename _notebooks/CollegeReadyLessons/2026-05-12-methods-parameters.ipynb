<summary>Methods and Parameters</summary>

A method is a function that belongs to an object or class. Parameters are named inputs that make a method flexible. The same method can run with different arguments, which keeps code reusable instead of hardcoded.

<pre><code class="language-javascript">class Greeter {
  greet(name) {
    console.log("Hello, " + name + "!");
  }
}

const greeter = new Greeter();
greeter.greet("Ava");</code></pre>
