## Popcorn Hack 2: Sum and Average

Prediction: The total volunteer hours should be `42`, and the average should be `7.0`. The empty-list guard prevents division by zero.

````python
# CODE_RUNNER: Lists Popcorn 2

volunteer_hours = [7, 6, 8, 5, 7, 9]

total_hours = 0
for hours in volunteer_hours:
    total_hours = total_hours + hours

if len(volunteer_hours) == 0:
    print("No data yet")
else:
    average_hours = total_hours / len(volunteer_hours)
    print(total_hours)
    print(average_hours)
````

## Popcorn Hack 3: Count Then Filter

Prediction: `socks` has 3 votes and `meals` has 2, so those are the popular items with two or more votes.

````python
# CODE_RUNNER: Lists Popcorn 3 - count then filter

items_requested = ["blankets", "socks", "meals", "socks", "water", "socks", "meals"]

frequency = {}
for item in items_requested:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1

popular = []
for item in frequency:
    if frequency[item] >= 2:
        popular.append(item)

print(frequency)
print(popular)
````

## Homework Hack: Donations Dataset

This combines the three required patterns on one dataset: indexing, accumulator total and average, frequency count, and filtering.

````python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter on one dataset

donations_by_item = ["blankets", "socks", "blankets", "water", "socks", "blankets",
                     "meals", "socks", "blankets", "water", "blankets", "socks"]

costs = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# LESSON 1 - indexing
print(len(donations_by_item))
print(donations_by_item[-1])

# LESSON 2 - sum and average
total = 0
for cost in costs:
    total = total + cost

if len(costs) == 0:
    print("No cost data")
else:
    average = total / len(costs)
    print("Total:", int(total))
    print("Average:", int(average))

# LESSON 3 - count and filter
frequency = {}
for item in donations_by_item:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1

popular = []
for item in frequency:
    if frequency[item] >= 3:
        popular.append(item)

print(frequency)
print(popular)
````

Validation notes: the last donation is found with `[-1]`, not a hardcoded positive index. The total uses an explicit accumulator loop instead of `sum()`. The filter builds a new `popular` list, leaving `donations_by_item` unchanged.
