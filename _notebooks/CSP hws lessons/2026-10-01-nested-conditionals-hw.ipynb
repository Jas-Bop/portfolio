## Popcorn Hack 1: After-School Homework Plan

Prediction: With `is_weekday = True` and `homework_finished = False`, the program enters the weekday branch, then the unfinished-homework branch, so it prints `Finish homework`.

````python
# CODE_RUNNER: Popcorn Hack 1 - Test the after-school homework plan

is_weekday = True
homework_finished = False

if is_weekday:
    if homework_finished:
        print("Free time")
    else:
        print("Finish homework")
else:
    print("Weekend plan")
````

Path followed: outer branch `is_weekday` is true, inner branch `homework_finished` is false.

Other tested paths:

- `is_weekday = True`, `homework_finished = True` -> `Free time`
- `is_weekday = False`, any homework value -> `Weekend plan`

## Popcorn Hack 2: Repair the Student Account Check

The original code was not nested because `if has_permission:` lined up with `if logged_in:`. That means Python checked permission even when the student was not logged in. The fixed version only checks permission inside the logged-in branch.

````python
# CODE_RUNNER: Popcorn Hack 2 - Repair the student account check

logged_in = True
has_permission = False

if logged_in:
    print("Account found")
    if has_permission:
        print("Page opened")
    else:
        print("Permission needed")
else:
    print("Log in first")
````

Tested login paths:

- `logged_in = True`, `has_permission = True` -> `Account found`, then `Page opened`
- `logged_in = True`, `has_permission = False` -> `Account found`, then `Permission needed`
- `logged_in = False`, any permission value -> `Log in first`

## Popcorn Hack 3: Lunch and Sports Practice

The outer condition checks whether lunch is packed first. Inside each lunch path, the inner condition checks whether practice happens after school.

````python
# CODE_RUNNER: Popcorn Hack 3 - Test the lunch and sports-practice plan

lunch_packed = True
practice_after_school = True

if lunch_packed:
    if practice_after_school:
        print("Packed lunch ready. Bring an extra snack for practice.")
    else:
        print("Packed lunch ready. No extra snack needed today.")
else:
    if practice_after_school:
        print("Buy lunch and save money for an after-practice snack.")
    else:
        print("Buy lunch. Head home after school.")
````

Four tested paths:

- `True`, `True` -> packed lunch plus extra snack
- `True`, `False` -> packed lunch, no extra snack
- `False`, `True` -> buy lunch and save snack money
- `False`, `False` -> buy lunch and head home

## Compact Decision Tree

```text
logged_in?
|-- no  -> Log in first
`-- yes -> homework_finished?
          |-- no  -> Finish homework before schedule choices
          `-- yes -> lunch_packed?
                    |-- yes -> practice_today?
                    |         |-- yes -> Packed lunch + bring practice snack
                    |         `-- no  -> Packed lunch + free afternoon
                    `-- no  -> practice_today?
                              |-- yes -> Buy lunch + save snack money
                              `-- no  -> Buy lunch + head home
```

This tree makes the nesting order visible: the schedule checks only matter after the student logs in, and lunch/practice choices only happen after homework is finished.

## Homework Hack: High School Daily Routine Dashboard

The dashboard stores account access, homework status, lunch status, and sports-practice status. It uses an outer account decision before checking the rest of the schedule.

````python
# CODE_RUNNER: Homework Hack - Build a high school daily routine dashboard

student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True
}

if student["logged_in"]:
    print("Dashboard opened")
    if student["homework_finished"]:
        if student["lunch_packed"]:
            if student["practice_today"]:
                print("Homework done. Bring your packed lunch and an extra snack for practice.")
            else:
                print("Homework done. Bring your packed lunch and enjoy free time after school.")
        else:
            if student["practice_today"]:
                print("Homework done. Buy lunch and save money for an after-practice snack.")
            else:
                print("Homework done. Buy lunch and head home after school.")
    else:
        print("Finish homework before choosing dinner, sports, or free time.")
else:
    print("Log in first")
````

Explanation: The schedule recommendations depend on the student being logged in because the dashboard should not show personal school-day advice before account access is confirmed. The lunch and practice checks depend on homework being finished because the program prioritizes required work before optional after-school plans.
