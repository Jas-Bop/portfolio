## Popcorn Hack 2: Backend Create Rule

Prediction: The first record should be rejected because spec number `1.1` already exists. The changed record with spec number `1.3` should be accepted.

````python
# CODE_RUNNER: Popcorn Hack 2 - Backend Create Rule

valid_categories = ["Auto Racing", "Drag Racing"]
existing_spec_numbers = ["1.1", "2.1"]

part = {
    "product_name": "Replacement Flywheels and Clutch Assemblies",
    "category": "Auto Racing",
    "spec_number": "1.1",
    "effective_date": "Nov. 9, 2001"
}

has_required_fields = (
    part["product_name"] != ""
    and part["spec_number"] != ""
    and part["effective_date"] != ""
)
valid_category = part["category"] in valid_categories
duplicate_spec = part["spec_number"] in existing_spec_numbers
can_create_record = has_required_fields and valid_category and not duplicate_spec

print("Can create original record:", can_create_record)

part["spec_number"] = "1.3"
duplicate_spec = part["spec_number"] in existing_spec_numbers
can_create_record = has_required_fields and valid_category and not duplicate_spec

print("Can create changed record:", can_create_record)
````

Actual output:

```text
Can create original record: False
Can create changed record: True
```

## Popcorn Hack 3: SFI Search Filter

Prediction: Query `1.2` should match only the Drag Racing record with spec number `1.2`. The other records are accepted categories, but they do not match the query.

````python
# CODE_RUNNER: Popcorn Hack 3 - SFI Search Filter

query = "1.2"

records = [
    {
        "product_name": "Replacement Flywheels and Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

valid_categories = ["Auto Racing", "Drag Racing"]

for record in records:
    query_matches = query.lower() in record["product_name"].lower() or query == record["spec_number"]
    category_is_valid = record["category"] in valid_categories
    include_record = query_matches and category_is_valid

    if include_record:
        print(record["product_name"], "-", record["spec_number"])

print("Search complete")
````

Actual output:

```text
Multiple Disc Clutch Assemblies - 1.2
Search complete
```

## Homework Hack: SFI Backend Validator

A valid record has a non-empty product name, an accepted racing category, a non-empty spec number, an effective date, and a spec number that does not already exist in the backend.

````python
# CODE_RUNNER: Homework Hack - SFI Backend Validator

existing_spec_numbers = ["1.1", "2.1"]
valid_categories = ["Auto Racing", "Drag Racing"]

test_records = [
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.2",
        "effective_date": "Feb. 9, 2006"
    },
    {
        "product_name": "Replacement Flywheels",
        "category": "Street Car",
        "spec_number": "3.1",
        "effective_date": "Jan. 1, 2026"
    },
    {
        "product_name": "Existing Flywheel Record",
        "category": "Drag Racing",
        "spec_number": "1.1",
        "effective_date": "Jan. 1, 2026"
    },
    {
        "product_name": "",
        "category": "Drag Racing",
        "spec_number": "4.1",
        "effective_date": "Jan. 1, 2026"
    }
]

def validate_sfi_record(record):
    has_product_name = record["product_name"] != ""
    valid_category = record["category"] in valid_categories
    has_spec_number = record["spec_number"] != ""
    has_effective_date = record["effective_date"] != ""
    duplicate_spec = record["spec_number"] in existing_spec_numbers

    is_valid = (
        has_product_name
        and valid_category
        and has_spec_number
        and has_effective_date
        and not duplicate_spec
    )

    return is_valid

for record in test_records:
    if validate_sfi_record(record):
        print("ACCEPT:", record["spec_number"], record["product_name"])
    else:
        print("REJECT:", record["spec_number"], record["product_name"])
````

Actual output:

```text
ACCEPT: 1.2 Multiple Disc Clutch Assemblies
REJECT: 3.1 Replacement Flywheels
REJECT: 1.1 Existing Flywheel Record
REJECT: 4.1 
```

Boolean logic explanation: the validator builds small Boolean checks first, then combines them with `and` because every requirement must pass. It uses `not duplicate_spec` because a record is only valid when the spec number is new.
