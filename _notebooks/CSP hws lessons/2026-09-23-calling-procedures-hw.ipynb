```python
# CODE_RUNNER: HOMEWORK HACK

def check_priority(incident_type):
    # TODO: Return True when the incident requires priority handling
    priority_incidents = ["Medical Emergency", "Fire", "Gas Leak"]
    return incident_type in priority_incidents


def dispatch_response(location, incident_type):
    # TODO: Call check_priority()
    # TODO: Store the returned value in a variable
    needs_priority = check_priority(incident_type)

    # TODO: Use an if/else statement to print the appropriate response
    if needs_priority:
        print(f"Dispatch priority response to {location} for: {incident_type}")
    else:
        print(f"Dispatch standard response to {location} for: {incident_type}")


# TODO: Call dispatch_response() at least three times
dispatch_response("Poway", "Medical Emergency")
dispatch_response("San Diego", "Road Closure")
dispatch_response("Escondido", "Fire")
```
