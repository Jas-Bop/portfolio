## Popcorn Hack: Wildfire Dispatch Scanner

Scan wildfire alert levels, skip safe zones, and stop at the first critical zone. My safe cutoff is 4 and my critical cutoff is 9. A critical alert also counts as an action alert.


```python
# CODE_RUNNER: Popcorn Hack - wildfire dispatch scanner

alerts = [2, 5, 1, 7, 4, 9, 6]
safe_cutoff = 4
critical_cutoff = 9

checked_count = 0
action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

for alert in alerts:
    checked_count += 1
    if alert < safe_cutoff:
        safe_count += 1
        print(f"Safe wildfire alert {alert}: no crew needed")
        continue

    action_count += 1
    total_action_score += alert
    print(f"Dispatch a crew for wildfire alert {alert}")

    if alert >= critical_cutoff:
        critical_found = True
        print("Critical wildfire alert: stop scanning and focus the response")
        break

print("\nWildfire dispatch final report")
print("Checked alerts:", checked_count)
print("Safe alerts skipped:", safe_count)
print("Action alerts:", action_count)
print("Total action score:", total_action_score)
print("Critical found:", "yes" if critical_found else "no")
```

## Popcorn Explanation

The `for` loop reads each alert in order. `checked_count` increases before the safe check so skipped alerts still count as checked. `continue` skips the dispatch code for safe alerts. The action counter and total update before the critical check so the critical alert is included. `break` ends the scan at 9, leaving the last alert, 6, unchecked.

## Homework Hack: Server Load Monitor

My checker monitors ten server CPU-load percentages. Loads below 60% are safe, loads from 60% to below 90% need attention, and loads at or above 90% are critical. The critical load is included in the action count and total before the scan stops.


```python
# CODE_RUNNER: Homework Hack - server CPU load checker

server_loads = [28, 60, 45, 76, 59, 84, 90, 67, 38, 95]
safe_cutoff = 60
critical_cutoff = 90

checked_count = 0
action_count = 0
skipped_count = 0
total_score = 0
critical_found = False

for server_load in server_loads:
    checked_count += 1
    print(f"Server {checked_count}: CPU load {server_load}%")

    if server_load < safe_cutoff:
        skipped_count += 1
        print("  Safe load: skip intervention")
        continue

    action_count += 1
    total_score += server_load
    print("  Action needed: investigate server load")

    if server_load >= critical_cutoff:
        critical_found = True
        print("  Critical CPU load: stop scanning and respond immediately")
        break

print("\nServer load final report")
print("Readings available:", len(server_loads))
print(f"Thresholds: action >= {safe_cutoff}%, critical >= {critical_cutoff}%")
print("Checked count:", checked_count)
print("Skipped count:", skipped_count)
print("Action count:", action_count)
print("Total action load score:", total_score)
print("Critical found:", "yes" if critical_found else "no")
print("Readings left unchecked:", len(server_loads) - checked_count)
```

`continue` skips only the current safe reading. `break` stops the entire loop, so the final three readings are not checked. The 60% reading is exactly at the action threshold and requires action; the 90% reading is exactly at the critical threshold and stops the scan. The checked count equals the skipped count plus the action count: `7 = 3 + 4`.

A `for` loop fits this task because the readings are already in a list. A `while` loop would fit a task that repeatedly checks a changing condition, such as cooling equipment until its temperature is safe; its temperature must update each pass to avoid an infinite loop.
