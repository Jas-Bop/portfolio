## Popcorn

````python
# CODE_RUNNER: UESL 3.01 Popcorn - Keep the checkpoint
stars = 8
checkpoint_stars = stars
stars = stars + 4
print(stars)
print(checkpoint_stars)
````

Prediction: The live stars will print `12`, and the checkpoint stars will print `8`.

Actual output:

```text
12
8
```

Explanation: `checkpoint_stars` stores the value of `stars` when `stars` is still `8`. Later, `stars` is reassigned to `12`, but the checkpoint does not automatically change.

## Homework

````python
# CODE_RUNNER: UESL 3.1 Homework - Star counter
# Your UESL prototype goes here.
game_name = "UESL Star Dash"
reduced_motion = True

level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

stored_total = level_one_stars + level_two_stars + level_three_stars

print("Game:", game_name)
print("Reduced motion:", reduced_motion)
print("Original total:", stored_total)

bonus = 5
level_two_stars = level_two_stars + bonus

print("Stored total after bonus:", stored_total)

updated_total = level_one_stars + level_two_stars + level_three_stars
print("Updated total:", updated_total)
````

Prediction: The game name and reduced-motion setting will print first. The original total will be `40`. After adding a bonus of `5` to level two, the stored total will still be `40`, but the recalculated updated total will be `45`.

Actual output:

```text
Game: UESL Star Dash
Reduced motion: True
Original total: 40
Stored total after bonus: 40
Updated total: 45
```

## Design Thinking

Player need: A player needs the game to track stars clearly without exposing private player information.

Goal: My goal was to build a simple UESL star counter that shows how assignment, stored totals, and recalculation work.

Setting/representation considered: I considered using separate variables for each level because the next lesson will replace them with a collection. Separate variables are readable for three levels, but they would be hard to maintain for 100 levels.

Prototype built: I built a Python prototype with a game name, a reduced-motion Boolean, three level-star variables, a stored total, and a recalculated total after a bonus.

Test or revision: I tested the program with a `5` bonus and a `0` bonus. The `5` bonus proved that the stored total stays `40` until recalculated, while the `0` bonus proved that all totals remain `40` when no stars are added.

Reduced motion should be a player choice because some players may need fewer animations for comfort, accessibility, focus, or motion sensitivity.
