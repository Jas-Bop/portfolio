## POPCORN HACK 2


```python
# CODE_RUNNER: POPCORN HACK 2

risk_score = 9

# Broken version with risk_score >= 3 first prints the mild warning for 9.
# Unreachable in the broken version: risk_score >= 6 and risk_score >= 9.
# Fixed order: 9, 6, 3, else.
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

## HOMEWORK HACK


```python
# CODE_RUNNER: HOMEWORK HACK

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# Add to risk_score using conditionals, one signal at a time.
# Each if is independent because a message can trigger multiple signals at once.
# No else is needed, because we want to keep adding points for each matching signal.
if "gift card" in message:
    risk_score += 4
if amount_requested > 100:
    risk_score += 3
if not known_contact:
    risk_score += 2

# 4 thresholds -> 5 outcomes -> 5 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
elif risk_score >= 1:
    print("A little suspicious. Keep an eye on it.")
else:
    print("Nothing suspicious in this message.")

print(f"Final risk score: {risk_score}")
```
