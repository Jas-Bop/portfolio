# 3.6 Conditionals

A condition is a question that evaluates to True or False. Every condition must have all of its answers written down, and an `elif` chain must be ordered so no branch is unreachable.

This notebook includes the two Popcorn Hacks, the Homework Hack, a branch reachability proof, and the required extreme tests from the lesson.

## Popcorn Hack 1: Fix the silent failure

Run the cell first, then fix it so both outcomes print.

## Popcorn Hack 2: Fix the unreachable branch order

This version is not silent, but it is still wrong because some thresholds can never run. In the broken version, `risk_score >= 6` and `risk_score >= 9` can never run because every score that reaches them already matched `risk_score >= 3`.

````python
# CODE_RUNNER: 3.6 Popcorn 2 - Run it first. Then fix the order.

risk_score = 9

# Broken version with risk_score >= 3 first prints the mild warning for 9.
# Unreachable in the broken version: risk_score >= 6 and risk_score >= 9.
# Fixed order: 9, 6, 3, else.
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
````

## Homework Hack: Score the message and explain each branch

The score is built by checking one signal at a time. Each `if` adds points, but no `else` is needed because the code is meant to keep evaluating additional signals.

Use a separate `if` for each signal and then report the final score with an `if / elif / else` chain.

````python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# Add to risk_score using conditionals, one signal at a time.
# Each if is independent because a message can trigger multiple signals at once.
# No else is needed, because we want to keep adding points for each matching signal.
if "gift card" in message:
    risk_score += 4
if amount_requested > 100:
    risk_score += 3
if not known_contact:
    risk_score += 2

# 4 thresholds -> 5 outcomes -> 5 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
elif risk_score >= 1:
    print("A little suspicious. Keep an eye on it.")
else:
    print("Nothing suspicious in this message.")

print(f"Final risk score: {risk_score}")
````

### Reachability proof

One input set that reaches each answer in the reporting chain:

- `risk_score >= 9` -> message asks for gift cards, amount > 100, and sender is not known
- `risk_score >= 6` -> message asks for gift cards and amount > 100, but sender is known
- `risk_score >= 3` -> message asks for gift cards only, or amount > 100 only
- `risk_score >= 1` -> a low-risk message with a small amount or one minor signal
- `else` -> a clean message with no suspicious signals

- `risk_score >= 9` -> message asks for gift cards, amount is 500, and sender is not known. Score: 4 + 3 + 2 = 9.
- `risk_score >= 6` -> message asks for gift cards, amount is 500, and sender is known. Score: 4 + 3 = 7.
- `risk_score >= 3` -> message asks for gift cards only, amount is 0, and sender is known. Score: 4.
- `risk_score >= 1` -> message does not ask for gift cards, amount is 0, and sender is not known. Score: 2.
- `else` -> clean message, amount is 0, and sender is known. Score: 0.

The chain is reachable because each range is distinct and ordered from highest to lowest. The scoring `if` statements do not need `else` because they are independent signals: a message can trigger several at once, and a false signal should simply add 0 while the other signals still get checked.

Stretch explanation: an `if` with no `else` is fine during scoring because silence means "add no points for this one signal," but it is not fine in the final trainer decision because every final outcome must print advice.

## Extreme Tests

Both tests start from fresh input values and must print a sensible message.

````python
# CODE_RUNNER: 3.6 Test - clean message

message = "Hi Mom, are we still on for Sunday?"
amount_requested = 0
known_contact = True

risk_score = 0

# Each if is an independent signal, so false signals simply add 0.
if "gift card" in message:
    risk_score += 4
if amount_requested > 100:
    risk_score += 3
if not known_contact:
    risk_score += 2

# 4 thresholds -> 5 outcomes -> 5 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
elif risk_score >= 1:
    print("A little suspicious. Keep an eye on it.")
else:
    print("Nothing suspicious in this message.")

print(f"Final risk score: {risk_score}")
````
